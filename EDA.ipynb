```python
# cretaing the timebased eda suibg tehe started at column
final_df["started_at"]=pd.to_datetime(final_df["started_at"],errors="coerce")
final_df["started_hour"]=final_df["started_at"].dt.hour
final_df["started_day_of_week"]=final_df["started_at"].dt.day_name()
final_df["started_date"]=final_df["started_at"].dt.date
```

```python
hourly_rides=final_df["started_hour"].value_counts().sort_index()

plt.figure(figsize=(10,5))

sns.barplot(
    x=hourly_rides.index,
    y=hourly_rides.values,
    palette="mako"
)
plt.title("Numbeer of rides by hour")
plt.xlabel("Hour")
plt.ylabel("Number of rides")
plt.show()
```

```python
# rides by the day of the week
days=[
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]
daily_ride=(final_df["started_day_of_week"].value_counts().reindex(days))

plt.figure(figsize=(10,5))
sns.barplot(
    x=daily_ride.index,
    y=daily_ride.values,
    palette="viridis"
)

plt.title("No. of rides by Days")
plt.xlabel("Days")
plt.ylabel("no. of rides")
plt.show()
```

```python
# rides by the date
date=final_df["started_date"].value_counts().sort_index()

plt.figure(figsize=(10,5))

plt.plot(
    date.index,
    date.values,
    color="maroon",
    marker="*",
    markerfacecolor="yellow"
)
plt.title('Daily ride count')
plt.xlabel("Date")
plt.ylabel("No. of rides")
plt.show()
```

```python
# member casula by the hour column
hourly_ride=(final_df.groupby(["started_hour","member_casual"]).size().reset_index(name="ride_count"))

plt.figure(figsize=(12,5))

sns.lineplot(
    data=hourly_ride,
    x="started_hour",
    y='ride_count',
    hue="member_casual",
    palette="plasma",
    marker="8",
    markerfacecolor="black"

)
plt.title("Member vs Casual hourly rides")
plt.xlabel("Hours")
plt.ylabel("No. of rides")
plt.show()
```

```python
# Memvbe casual vs day of the week

day_rides=(final_df.groupby(["started_day_of_week",'member_casual']).size().reset_index(name="ride_count"))

day_rides["started_day_of_week"]=pd.Categorical(day_rides["started_day_of_week"],categories=days,ordered=True)

day_rides=day_rides.sort_values("started_day_of_week")

plt.figure(figsize=(11,5))

sns.barplot(
    data=day_rides,
    x="started_day_of_week",
    y="ride_count",
    hue="member_casual",
    palette="magma"
)
plt.title("Member vs Casual Rides by Days")
plt.xlabel("Days")
plt.ylabel("No. of Rides")
plt.show()
```

```python
# weather vs no. of day_rides

weather_rides=(final_df.groupby("time").agg(
    ride_count=("ride_id","count"),
    temperature=("temperature_2m","first"),
    humidity=("relative_humidity_2m", "first"),
    precipitation=("precipitation", "first"),
    wind_speed=("wind_speed_10m", "first")
).reset_index())
```

```python
# temp vs rides

plt.figure(figsize=(9,5))

sns.scatterplot(
    data=weather_rides,
    x="temperature",
    y="ride_count",
    alpha=0.5,
    palette="rocket"
)
plt.title("Temperature vs the Rides")
plt.xlabel("Temperature")
plt.ylabel("No. of rides")
plt.show()
```

```python
# Percipitationvs rides

plt.figure(figsize=(7,5))

sns.scatterplot(
    data=weather_rides,
    x="precipitation",
    y="ride_count",
    alpha=0.5,
    palette="rocket"
)
plt.title("Precipitation vs the Rides")
plt.xlabel("Precipitation")
plt.ylabel("No. of rides")
plt.show()
```

```python
# wind speed vs rides

plt.figure(figsize=(7,5))

sns.scatterplot(
    data=weather_rides,
    x="wind_speed",
    y="ride_count",
    alpha=0.5,
    palette="rocket"
)
plt.title("Wind speed vs the Rides")
plt.xlabel("Wind speed")
plt.ylabel("No. of rides")
plt.show()
```

```python
# weather And the rider type

weather_rider=(
    final_df.groupby(["time","member_casual"])
    .agg(
        ride_count=("ride_id","count"),
        temperature=("temperature_2m","first"),
        precipitation=("precipitation","first")

    ).reset_index()
)

# visualisation

plt.figure(figsize=(7,5))

sns.scatterplot(
    data=weather_rider,
    x="temperature",
    y="ride_count",
    hue="member_casual",
    alpha=0.6,
    palette="rocket"
)

plt.title("Temp vs rides by rider type")
plt.xlabel("temperatuer")
plt.ylabel("No. of rides")
plt.show()
```
